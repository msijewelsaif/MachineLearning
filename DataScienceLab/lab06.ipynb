```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python


def analyzer(max_depth):
    # Load dataset
    data = pd.read_csv('/content/drive/MyDrive/DS LAB/After_Mid/Social_Network_Ads.csv')

    # Encode Gender column
    data['Gender'].replace({'Male': 0, 'Female': 1}, inplace=True)

    # Features and target selection
    x = data.iloc[:, 1:3].values  # Using first two features for visualization
    y = data.iloc[:, -1].values

    # Train Decision Tree Classifier
    clf4 = DecisionTreeClassifier(max_depth=max_depth)
    clf4.fit(x, y)

    # Create meshgrid for visualization
    a = np.arange(start=x[:, 0].min() - 1, stop=x[:, 0].max() + 1, step=0.1)
    b = np.arange(start=x[:, 1].min() - 1, stop=x[:, 1].max() + 1, step=0.1)
    xx, yy = np.meshgrid(a, b)

    # Predict labels for meshgrid points
    input_array = np.array([xx.ravel(), yy.ravel()]).T
    labels = clf4.predict(input_array)

    # Plot decision boundary
    plt.contourf(xx, yy, labels.reshape(xx.shape), alpha=0.75, cmap=plt.cm.viridis)
    plt.scatter(x[:, 0], x[:, 1], c=y, s=75, edgecolors='black', cmap=plt.cm.viridis)

```

```python
analyzer(max_depth=None)
```

```python
analyzer(max_depth=1)
```

```python
analyzer(max_depth=2)
```

```python
analyzer(max_depth=3)
```

```python
analyzer(max_depth=4)
```

```python
analyzer(max_depth=5)
```

```python
x=data.iloc[:,1:4].values
y=data.iloc[:,-1].values
```

```python
from sklearn.preprocessing import StandardScaler
scaller=StandardScaler()
x=scaller.fit_transform(x)
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf5=DecisionTreeClassifier()
clf5.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred=clf5.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
param_dist={ "criterion":['gini','entropy'],
            "max_depth":[1.2,3,4,5,None]}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(clf5,param_grid=param_dist,cv=10,n_jobs=-1)
```

```python
grid.fit(x_train,y_train)
```
