```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Generate a synthetic dataset
# Creating a random dataset for 100 samples (features: Age, BMI, Glucose Level, etc.)
np.random.seed(0)

# Simulate some features
data = {
    'Age': np.random.randint(20, 60, size=100),  # Age between 20 and 60
    'BMI': np.random.uniform(18.5, 35.0, size=100),  # BMI between 18.5 and 35
    'Glucose': np.random.randint(70, 200, size=100),  # Glucose level between 70 and 200
    'BloodPressure': np.random.randint(60, 120, size=100),  # Blood pressure between 60 and 120
    'Insulin': np.random.randint(5, 50, size=100),  # Insulin level between 5 and 50
    'DiabetesPedigreeFunction': np.random.uniform(0.1, 1.0, size=100),  # Genetic predisposition score
    'Outcome': np.random.choice([0, 1], size=100)  # Outcome (0: No Diabetes, 1: Diabetes)
}
```

```python
# Create a DataFrame
df = pd.DataFrame(data)

# 2. Data Preprocessing (Handle missing values, if any)
# For simplicity, we assume the dataset doesn't have missing values in this case.

# 3. Define Features and Target
X = df.drop(columns=['Outcome'])
y = df['Outcome']
```

```python
# 4. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
# 5. Initialize and train Logistic Regression model
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
# 6. Make predictions on the test set
y_pred = model.predict(X_test)
```

```python
# 7. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

# Displaying Results
print(f"Accuracy: {accuracy:.2f}")
print("\nConfusion Matrix:")
print(conf_matrix)
print("\nClassification Report:")
print(class_report)
```

```python
# 8. Predicting diabetes based on user input

def predict_diabetes(age, bmi, glucose, blood_pressure, insulin, pedigree):
    # Create a DataFrame from the user input
    input_data = pd.DataFrame({
        'Age': [age],
        'BMI': [bmi],
        'Glucose': [glucose],
        'BloodPressure': [blood_pressure],
        'Insulin': [insulin],
        'DiabetesPedigreeFunction': [pedigree]
    })

    # Make prediction
    prediction = model.predict(input_data)

    # Output result
    if prediction == 1:
        return "The person is likely to have diabetes."
    else:
        return "The person is unlikely to have diabetes."

# 9. Interactive input from the user
print("\nPlease enter the following details to predict diabetes:")
age = int(input("Age: "))
bmi = float(input("BMI: "))
glucose = int(input("Glucose Level: "))
blood_pressure = int(input("Blood Pressure: "))
insulin = int(input("Insulin Level: "))
pedigree = float(input("Diabetes Pedigree Function (Genetic Predisposition Score): "))

result = predict_diabetes(age, bmi, glucose, blood_pressure, insulin, pedigree)
print("\nPrediction:", result)
```
